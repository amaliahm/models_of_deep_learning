# Torchvision Image Transformation Examples

## Libraries

```python
import torch
import numpy as np
import PIL
from torch.utils.data import Dataset
from torch.utils.data import DataLoader
from torchvision import datasets
from torchvision import transforms
import matplotlib.pyplot as plt
from PIL import Image
```

```python
RANDOM_SEED = 123
BATCH_SIZE = 10

if torch.cuda.is_available():
    torch.backends.cudnn.deterministic = True
```

## CIFAR-10 Dataset

# 1) No Augmentation

```python
custom_transform = transforms.Compose([transforms.ToTensor()])

train_dataset = datasets.CIFAR10(root='data', 
                                 train=True, 
                                 transform=custom_transform,
                                 download=True)

train_loader = DataLoader(dataset=train_dataset, 
                          batch_size=BATCH_SIZE, 
                          num_workers=1,
                          shuffle=True)

torch.manual_seed(RANDOM_SEED)
for images, labels in train_loader:  
    images1 = images
    break

fig, axs = plt.subplots(1, 5, figsize=(9, 3), sharey=True)
for i in range(5):
    axs[i].imshow(np.transpose(images1[i], (1, 2, 0)))
```

# 2) Horizontal Flip

```python
custom_transform = transforms.Compose([
    transforms.RandomHorizontalFlip(p=0.5),
    transforms.ToTensor()])

train_dataset = datasets.CIFAR10(root='data', 
                                 train=True, 
                                 transform=custom_transform,
                                 download=False)

train_loader = DataLoader(dataset=train_dataset, 
                          batch_size=BATCH_SIZE, 
                          num_workers=1,
                          shuffle=True)

torch.manual_seed(RANDOM_SEED)
for images, labels in train_loader:  
    images2 = images
    break

fig, axs = plt.subplots(1, 5, figsize=(9, 3), sharey=True)
for i in range(5):
    axs[i].imshow(np.transpose(images2[i], (1, 2, 0)))
```

# 3) Random Rotation

```python
from torchvision import transforms
from PIL import Image

custom_transform = transforms.Compose([
    transforms.RandomRotation(
        degrees=20,
        interpolation=transforms.InterpolationMode.BILINEAR  # <- updated
    ),
    transforms.ToTensor()
])

train_dataset = datasets.CIFAR10(root='data', 
                                 train=True, 
                                 transform=custom_transform,
                                 download=False)

train_loader = DataLoader(dataset=train_dataset, 
                          batch_size=BATCH_SIZE, 
                          num_workers=1,
                          shuffle=True)

torch.manual_seed(RANDOM_SEED)
for images, labels in train_loader:  
    images2 = images
    break

fig, axs = plt.subplots(1, 5, figsize=(9, 3), sharey=True)
for i in range(5):
    axs[i].imshow(np.transpose(images2[i], (1, 2, 0)))
```

# 4) Normalize to [-1, 1] range

```python
custom_transform = transforms.Compose([
    transforms.ToTensor(),
    transforms.Normalize(std=(0.5, 0.5, 0.5), 
                         mean=(0.5, 0.5, 0.5))
])

train_dataset = datasets.CIFAR10(root='data', 
                                 train=True, 
                                 transform=custom_transform,
                                 download=False)

train_loader = DataLoader(dataset=train_dataset, 
                          batch_size=BATCH_SIZE, 
                          num_workers=1,
                          shuffle=True)

torch.manual_seed(RANDOM_SEED)
for images, labels in train_loader:  
    images2 = images
    break
    
print('Min-Max:', images2[0].min(), images2[0].max())
```

# 5) Random Color Jitter

```python
custom_transform = transforms.Compose([
    transforms.ColorJitter(brightness=(0.75, 1.25),
                           ),
    transforms.ToTensor(),
])

train_dataset = datasets.CIFAR10(root='data', 
                                 train=True, 
                                 transform=custom_transform,
                                 download=False)

train_loader = DataLoader(dataset=train_dataset, 
                          batch_size=BATCH_SIZE, 
                          num_workers=1,
                          shuffle=True)

torch.manual_seed(RANDOM_SEED)
for images, labels in train_loader:  
    images2 = images
    break

fig, axs = plt.subplots(1, 5, figsize=(9, 3), sharey=True)
for i in range(5):
    axs[i].imshow(np.transpose(images2[i], (1, 2, 0)))
    
plt.show()
```

# 6) Resize and Random Cropt

```python
custom_transform = transforms.Compose([
    transforms.Resize((38, 38)),
    transforms.RandomCrop((32, 32)),
    transforms.ToTensor(),
])

train_dataset = datasets.CIFAR10(root='data', 
                                 train=True, 
                                 transform=custom_transform,
                                 download=False)

train_loader = DataLoader(dataset=train_dataset, 
                          batch_size=BATCH_SIZE, 
                          num_workers=1,
                          shuffle=True)

torch.manual_seed(RANDOM_SEED)
for images, labels in train_loader:  
    images2 = images
    break

fig, axs = plt.subplots(1, 5, figsize=(9, 3), sharey=True)
for i in range(5):
    axs[i].imshow(np.transpose(images2[i], (1, 2, 0)))
    
plt.show()
```
